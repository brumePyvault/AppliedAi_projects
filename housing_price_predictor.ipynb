```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
from sklearn.model_selection import train_test_split
import pandas as pd
```

```python
house_csv='./data/house-prices-advanced-regression-techniques/train.csv'

house_data = pd.read_csv(house_csv)
```

```python
house_data['MSSubClass'] = house_data['MSSubClass'].astype(str)

features = ['MSSubClass', 'MSZoning', 'LotFrontage', 'LotArea', 'Street',
       'Alley', 'LotShape', 'LandContour', 'Utilities', 'LotConfig',
       'LandSlope', 'Neighborhood', 'Condition1', 'Condition2', 'BldgType',
       'HouseStyle', 'OverallQual', 'OverallCond', 'YearBuilt', 'YearRemodAdd',
        'Exterior1st', 'Exterior2nd', 'MasVnrType',
       'MasVnrArea', 'ExterQual', 'ExterCond', 'Foundation', 'BsmtQual',
       'BsmtCond', 'BsmtExposure', 'BsmtFinType1', 'BsmtFinSF1',
       'BsmtFinType2', 'BsmtFinSF2', 'BsmtUnfSF', 'TotalBsmtSF', 'Heating',
       'HeatingQC', 'CentralAir', '1stFlrSF', '2ndFlrSF',
       'LowQualFinSF', 'GrLivArea', 'BsmtFullBath', 'BsmtHalfBath', 'FullBath',
       'HalfBath', 'BedroomAbvGr', 'KitchenAbvGr', 'KitchenQual',
       'TotRmsAbvGrd', 'Functional', 'Fireplaces', 'FireplaceQu', 'GarageType',
       'GarageYrBlt', 'GarageFinish', 'GarageCars', 'GarageArea', 'GarageQual',
       'GarageCond', 'PavedDrive', 'WoodDeckSF', 'OpenPorchSF',
       'EnclosedPorch', '3SsnPorch', 'ScreenPorch', 'PoolArea', 'PoolQC',
       'Fence', 'MiscFeature', 'MiscVal', 'MoSold', 'YrSold', 'SaleType',
       'SaleCondition'
       ]

# Step 3: One-hot encode categorical features (MSSubClass)
X = pd.get_dummies(house_data[features])
y = house_data['SalePrice']

# Step 4: Train-test split
X_train, X_valid, y_train, y_valid = train_test_split(X, y, random_state=42)

# Step 5: Fit RandomForest
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# Step 6: Evaluate
predictions = model.predict(X_valid)
mae = mean_absolute_error(y_valid, predictions)
print(f"Mean Absolute Error: {mae:,.2f}")
```

```python
# Load the test data
test_csv = './data/house-prices-advanced-regression-techniques/test.csv'
test_data = pd.read_csv(test_csv)

# Preserve Ids for submission
test_ids = test_data['Id']

# Convert MSSubClass in test data
test_data['MSSubClass'] = test_data['MSSubClass'].astype(str)

# One-hot encode test features
X_test = pd.get_dummies(test_data[features])

# Align test set to match training features
X_test, _ = X_test.align(X_train, join='right', axis=1, fill_value=0)

# Predict
test_predictions = model.predict(X_test)

# Save predictions
submission = pd.DataFrame({
    'Id': test_data['Id'],
    'SalePrice': test_predictions
})

submission.to_csv('house_price_predictions.csv', index=False)
print("✅ Prediction file saved: house_price_predictions.csv")

```
